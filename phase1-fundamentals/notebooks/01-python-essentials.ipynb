## 1. Bien (Variable) la gi?

**Giai thich don gian:**

Ban co the hieu **bien** nhu mot **nhan dan** gan tren mot gia tri.

Vi du thuc te:
- Ban co mot hop, dan nhan "gia_vi" ben ngoai
- Ben trong hop la gia tri "75000" (gia co phieu)

Trong Python:
```python
gia_co_phieu = 75000
```

Tao mot "hop" ten `gia_co_phieu`, ben trong chua so `75000`

````python
# =============================================================================
# 1. Bien (Variable) la gi?
# =============================================================================

# Vi du 1: Luu ten cong ty
ten_cong_ty = "Vingroup"
print("Ten cong ty:", ten_cong_ty)

# Vi du 2: Luu gia co phieu
gia_co_phieu = 45000
print("Gia Vingroup:", gia_co_phieu, "VND")

# Vi du 3: Luu khoi luong giao dich
khoi_luong = 5_000_000  # 5 trieu co phieu
print("Khoi luong:", khoi_luong)

# Vi du 4: Tinh toan voi bien
so_co_phieu = 100
tri_gia = gia_co_phieu * so_co_phieu  # 45000 * 100
print("Tong gia tri:", tri_gia, "VND")
````

**Chu y quan trong:**

1. Ten bien **KHONG co khoang trang** - dung `_` thay the
   - ✅ `gia_co_phieu` 
   - ❌ `gia co phieu`

2. Python **phan biet hoa/thuong**
   - `Gia` ≠ `gia` ≠ `GIA`

3. Ten bien nen **co nghia** de de hieu
   - ✅ `gia_co_phieu` (ro rang)
   - ❌ `x` (khong biet la gi)

## 2. Cac loai du lieu (Data Types)

**Giai thich:**

Trong Python, moi gia tri co mot "loai" (type). Cac loai pho bien:

| Loai | Tieng Viet | Vi du |
|------|-----------|-------|
| `int` | So nguyen (khong co phan thap phan) | `25`, `100`, `1_000_000` |
| `float` | So thap phan (co phan thap phan) | `75.5`, `3.14`, `0.05` |
| `str` | Chu (van ban) | `"VNM"`, `"Vinamilk"` |
| `bool` | Dung hoac Sai | `True`, `False` |

````python
# =============================================================================
# 2. Cac loai du lieu
# =============================================================================

# --- int: So nguyen (khong co phan thap phan) ---
so_co_phieu = 100
nam = 2024
print("int - So nguyen:", so_co_phieu, "->", type(so_co_phieu))

# --- float: So thap phan ---
gia_tri = 75.5
ti_le = 0.05  # 5%
print("float - So thap phan:", gia_tri, "->", type(gia_tri))

# --- str: Chu (van ban) ---
ma_cp = "VNM"
ten_cty = "Vinamilk"
print("str - Chu:", ma_cp, "-", ten_cty, "->", type(ma_cp))

# --- bool: Dung hoac Sai ---
duoc_niem_phieu = True
bi_treo = False
print("bool - True/False:", duoc_niem_phieu, "->", type(duoc_niem_phieu))
````

**Kiem tra loai du lieu:**

Dung ham `type()` de xem loai cua mot bien:

```python
type(100)       # → int
type(75.5)      # → float
type("VNM")    # → str
type(True)      # → bool
```

## 3. Phep toan co ban

**Giai thich:**

Python co the lam cac phep toan nhu may tinh:

| Phep toan | Ky hieu | Vi du | Ket qua |
|-----------|---------|-------|--------|
| Cong | `+` | `10 + 5` | `15` |
| Tru | `-` | `10 - 5` | `5` |
| Nhan | `*` | `10 * 5` | `50` |
| Chia | `/` | `10 / 5` | `2.0` |
| Chia lay du | `%` | `10 % 3` | `1` |
| Luy thua | `**` | `10 ** 2` | `100` |

````python
# =============================================================================
# 3. Phep toan co ban
# =============================================================================

# Vi du: Tinh loi nhuan co phieu
gia_mua = 100000  # Gia mua: 100,000 VND
gia_ban = 120000  # Gia ban: 120,000 VND

# Tinh loi nhuan tuyet doi
loi_nhuan = gia_ban - gia_mua
print(f"Loi nhuan tuyet doi: {loi_nhuan:,} VND")

# Tinh ty suat loi nhuan (%)
ty_suat = (gia_ban - gia_mua) / gia_mua
print(f"Ty suat loi nhuan: {ty_suat:.2%}")  # .2% = hien thi 2 chu so thap phan

# Tinh gia tri sau khi tang 10%
gia_sau_tang = gia_mua * 1.10
print(f"Gia sau khi tang 10%: {gia_sau_tang:,.0f} VND")
````

**Meo format so:**

- `{:,.0f}` → hien thi so co dau phay: `120,000`
- `{:.2%}` → hien thi phan tram: `0.20` → `20.00%`
- `{:.2f}` → hien thi 2 so thap phan: `3.14`

## 4. List (Danh sach) - Luu nhieu gia tri

**Giai thich:**

List giong nhu mot **danh sach mua sam**. Ban co the luu nhieu thu trong mot bien.

```python
# Tao mot list (danh sach)
danh_sach = ["tao", "cam", "chuoi"]

# List co the chua so
gia_cac_loai = [75000, 45000, 35000]
```

````python
# =============================================================================
# 4. List (Danh sach)
# =============================================================================

# Vi du 1: Danh sach cac ma co phieu
cac_ma_cp = ["VNM", "HPG", "VIC", "MSN", "VHM"]
print("Danh sach ma CP:", cac_ma_cp)

# Vi du 2: Danh sach gia
gia_cac_cp = [75000, 35000, 45000, 85000, 38000]
print("Gia cac CP:", gia_cac_cp)

# --- Lay phan tu tu List ---
# CHI SO (INDEX) bat dau tu 0!
print()
print("Phan tu dau tien:", cac_ma_cp[0])   # VNM
print("Phan tu thu 2:", cac_ma_cp[1])    # HPG
print("Phan tu cuoi cung:", cac_ma_cp[-1]) # VHM
````

**QUAN TRONG: Chi so bat dau tu 0**

```
Danh sach:    ["A",    "B",    "C",    "D"]
Chi so:        0      1      2      3
              -4     -3     -2     -1   (dem tu cuoi)
```

Day la quy uoc cua hau het ngon ngu lap trinh. De nham lan luc dau, quen sau.

````python
# --- List Operations / Cac thao tac voi List ---

# Them phan tu vao cuoi
cac_ma_cp.append("FPT")
print("Sau khi them FPT:", cac_ma_cp)

# Dem so phan tu
print("So luong ma CP:", len(cac_ma_cp))

# Tim vi tri cua phan tu
vi_tri = cac_ma_cp.index("HPG")
print("HPG o vi tri so:", vi_tri)

# Kiem tra co trong list khong
co_fpt = "FPT" in cac_ma_cp
co_xyz = "XYZ" in cac_ma_cp
print(f"Co FPT trong danh sach? {co_fpt}")  # True
print(f"Co XYZ trong danh sach? {co_xyz}")  # False
````

## 5. Vong lap (Loops) - Lap di lap lai

**Giai thich:**

Vong lap giup ban **lam cung mot viec** nhieu lan.

Vi du trong doi thuc: "Gui tin nhan chuc mung den 100 khach hang"

Trong Python, co 2 loai vong lap:
- `for` — khi biet truoc so lan lap
- `while` — khi chi biet dieu kien dung

````python
# =============================================================================
# 5. Vong lap (Loops)
# =============================================================================

# --- for loop: Lap qua tung phan tu ---
print("=== Vi du 1: Duyet danh sach ===")
cac_ma_cp = ["VNM", "HPG", "VIC", "MSN"]

for ma_cp in cac_ma_cp:
    print(f"- {ma_cp}")

# Giai thich tung buoc:
# Buoc 1: ma_cp = "VNM", in ra "- VNM"
# Buoc 2: ma_cp = "HPG", in ra "- HPG"
# Buoc 3: ma_cp = "VIC", in ra "- VIC"
# Buoc 4: ma_cp = "MSN", in ra "- MSN"
# Het danh sach → dung
````

````python
# --- for loop voi range() ---
print("\n=== Vi du 2: Lap 5 lan ===")

for i in range(5):
    print(f"Lan thu {i + 1}")

# range(5) tao ra: 0, 1, 2, 3, 4
# Cong them 1 de hien thi: 1, 2, 3, 4, 5
````

````python
# --- while loop: Lap den khi dieu kien sai ---
print("\n=== Vi du 3: While loop ===")

so_du = 100000  # So du tai khoan
toc_do_tang = 1.10  # Tang 10% moi nam
nam = 0

while so_du < 200000:  # Lap den khi dat 200,000
    nam = nam + 1
    so_du = so_du * toc_do_tang
    print(f"Nam {nam}: {so_du:,.0f} VND")

print(f"\n→ Sau {nam} nam, so du dat {so_du:,.0f} VND")
````

## 6. Ham (Function) - Tai su dung code

**Giai thich:**

Ham giong nhu mot **cong thuc nau an**. Ban viet mot lan, dung nhieu lan.

```python
def ten_ham(tham_so):
    """Giai thich ham nay lam gi"""
    # Lam gi do voi tham_so
    return ket_qua
```

````python
# =============================================================================
# 6. Ham (Function)
# =============================================================================

# --- Vi du 1: Ham tinh loi nhuan ---
def tinh_loi_nhuan(gia_mua, gia_ban):
    """
    Tinh loi nhuan va ty suat loi nhuan
    
    Args:
        gia_mua: Gia mua (VND)
        gia_ban: Gia ban (VND)
    
    Returns:
        Tuple (loi nhuan tuyet doi, ty suat %)
    """
    loi_nhuan = gia_ban - gia_mua
    ty_suat = (gia_ban - gia_mua) / gia_mua
    return loi_nhuan, ty_suat

# Su dung ham
ln1, ts1 = tinh_loi_nhuan(100000, 120000)
print(f"Mua 100k, Ban 120k: Loi nhuan = {ln1:,} VND ({ts1:.1%})")

ln2, ts2 = tinh_loi_nhuan(75000, 69000)
print(f"Mua 75k, Ban 69k: Loi nhuan = {ln2:,} VND ({ts2:.1%})")
````

````python
# --- Vi du 2: Ham tinh gia tri danh muc ---
def tinh_gia_tri_danh_muc(hang_hoa):
    """
    Tinh tong gia tri danh muc dau tu
    
    Args:
        hang_hoa: Dict chua thong tin {ten: {gia, so_luong}}
    
    Returns:
        Tong gia tri
    """
    tong = 0
    for ten, thong_tin in hang_hoa.items():
        gia = thong_tin["gia"]
        so_luong = thong_tin["so_luong"]
        gia_tri = gia * so_luong
        tong = tong + gia_tri
        print(f"{ten}: {gia:,} x {so_luong} = {gia_tri:,} VND")
    
    return tong

# Tao danh muc dau tu
danh_muc = {
    "VNM": {"gia": 75000, "so_luong": 100},
    "HPG": {"gia": 35000, "so_luong": 200},
    "VIC": {"gia": 45000, "so_luong": 50}
}

print("=== Tinh gia tri danh muc ===")
tong_gia_tri = tinh_gia_tri_danh_muc(danh_muc)
print(f"\nTong gia tri: {tong_gia_tri:,} VND")
````

## 7. Dictionary - Luu du lieu dang cap Key-Value

**Giai thich:**

Dictionary giong nhu mot **danh ba dien thoai**:
- Key = Ten nguoi (dinh danh)
- Value = So dien thoai (thong tin)

```python
danh_ba = {
    "Minh": "0901234567",
    "Lan": "0912345678"
}
```

````python
# =============================================================================
# 7. Dictionary
# =============================================================================

# --- Vi du: Thong tin co phieu VNM ---
thong_tin_vnm = {
    "ma": "VNM",
    "ten": "Vinamilk",
    "gia": 75000,
    "khoi_luong": 5_000_000,
    "san": "HOSE"
}

# Lay thong tin
print("Ma:", thong_tin_vnm["ma"])
print("Ten:", thong_tin_vnm["ten"])
print("Gia:", thong_tin_vnm["gia"], "VND")

# Them thong tin moi
thong_tin_vnm["pe"] = 15.5  # P/E ratio
thong_tin_vnm["duoc_niem_phieu"] = True

print("P/E:", thong_tin_vnm["pe"])
print("Niem yet:", thong_tin_vnm["duoc_niem_phieu"])
````

````python
# --- Vi du: Nhieu co phieu ---
danh_sach_cp = {
    "VNM": {"gia": 75000, "kl": 5_000_000, "nganh": "Thuc pham"},
    "HPG": {"gia": 35000, "kl": 10_000_000, "nganh": "Thep"},
    "VIC": {"gia": 45000, "kl": 3_000_000, "nganh": "Bat dong san"}
}

# Duyet tung co phieu
print("\n=== Danh sach co phieu ===")
for ma, tt in danh_sach_cp.items():
    gia_tri = tt["gia"] * tt["kl"]
    print(f"{ma}: {tt['nganh']} - GT: {gia_tri:,.0f} VND")
````

## 8. Doc va Ghi File

**Giai thich:**

File la noi luu du lieu. Chung ta can:
- **Doc (Read)** — lay du lieu tu file
- **Ghi (Write)** — luu du lieu vao file

````python
# =============================================================================
# 8. Doc va Ghi File
# =============================================================================

import json  # Thu vien xu ly JSON
from pathlib import Path  # Quan ly duong dan

# --- Ghi du lieu vao file ---

# Tao du lieu mau
du_lieu = {
    "ngay": "2024-01-15",
    "danh_muc": [
        {"ma": "VNM", "gia": 75000, "so_luong": 100},
        {"ma": "HPG", "gia": 35000, "so_luong": 200}
    ]
}

# Duong dan file
duong_dan = Path("../data/danh_muc.json")
duong_dan.parent.mkdir(parents=True, exist_ok=True)

# Ghi file (nhu viet vao so)
with open(duong_dan, "w", encoding="utf-8") as f:
    json.dump(du_lieu, f, indent=2, ensure_ascii=False)

print(f"Da luu vao: {duong_dan}")
````

````python
# --- Doc du lieu tu file ---

# Doc file (nhu mo so ra doc)
with open(duong_dan, "r", encoding="utf-8") as f:
    du_lieu_doc = json.load(f)

print(f"Ngay: {du_lieu_doc['ngay']}")
print(f"So co phieu: {len(du_lieu_doc['danh_muc'])}")

for cp in du_lieu_doc["danh_muc"]:
    gia_tri = cp["gia"] * cp["so_luong"]
    print(f"  - {cp['ma']}: {gia_tri:,} VND")
````

**Meo:**

```python
# Luon dung "with open()" de doc/ghi file
# "with" dam bao file duoc dong lai dung cach
# encoding="utf-8" quan trong de xu ly tieng Viet
```
