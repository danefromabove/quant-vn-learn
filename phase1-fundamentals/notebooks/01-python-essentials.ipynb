## 1. Biến (Variable) là gì?

**Giải thích đơn giản:**

Bạn có thể hiểu **biến** như một **nhãn dán** gắn trên một giá trị.

Ví dụ thực tế:
- Bạn có một hộp, dán nhãn "giá_vi" bên ngoài
- Bên trong hộp là giá trị "75000" (giá cổ phiếu)

Trong Python:
```python
gia_co_phieu = 75000
```

→ Tạo một "hộp" tên `gia_co_phieu`, bên trong chứa số `75000`

````python
# =============================================================================
# 1. Biến (Variable) là gì?
# =============================================================================

# Ví dụ 1: Lưu tên công ty
ten_cong_ty = "Vingroup"
print("Tên công ty:", ten_cong_ty)

# Ví dụ 2: Lưu giá cổ phiếu
gia_co_phieu = 45000
print("Giá Vingroup:", gia_co_phieu, "VND")

# Ví dụ 3: Lưu khối lượng giao dịch
khoi_luong = 5_000_000  # 5 triệu cổ phiếu
print("Khối lượng:", khoi_luong)

# Ví dụ 4: Tính toán với biến
so_co_phieu = 100
tri_gia = gia_co_phieu * so_co_phieu  # 45000 * 100
print("Tổng giá trị:", tri_gia, "VND")
````

**Chú ý quan trọng:**

1. Tên biến **KHÔNG có khoảng trắng** → dùng `_` thay thế
   - ✅ `gia_co_phieu` 
   - ❌ `gia co phieu`

2. Python **phân biệt hoa/thường**
   - `Gia` ≠ `gia` ≠ `GIA`

3. Tên biến nên **có nghĩa** để dễ hiểu
   - ✅ `gia_co_phieu` (rõ ràng)
   - ❌ `x` (không biết là gì)

## 2. Các loại dữ liệu (Data Types)

**Giải thích:**

Trong Python, mỗi giá trị có một "loại" (type). Các loại phổ biến:

| Loại | Tiếng Việt | Ví dụ |
|------|-----------|-------|
| `int` | Số nguyên (không có phần thập phân) | `25`, `100`, `1_000_000` |
| `float` | Số thập phân (có phần thập phân) | `75.5`, `3.14`, `0.05` |
| `str` | Chữ (văn bản) | `"VNM"`, `"Vinamilk"` |
| `bool` | Đúng hoặc Sai | `True`, `False` |

````python
# =============================================================================
# 2. Các loại dữ liệu
# =============================================================================

# --- int: Số nguyên (không có phần thập phân) ---
so_co_phieu = 100
nam = 2024
print("int - Số nguyên:", so_co_phieu, "->", type(so_co_phieu))

# --- float: Số thập phân ---
gia_tri = 75.5
ti_le = 0.05  # 5%
print("float - Số thập phân:", gia_tri, "->", type(gia_tri))

# --- str: Chữ (văn bản) ---
ma_cp = "VNM"
ten_cty = "Vinamilk"
print("str - Chữ:", ma_cp, "-", ten_cty, "->", type(ma_cp))

# --- bool: Đúng hoặc Sai ---
duoc_niem_phieu = True
bi_treo = False
print("bool - True/False:", duoc_niem_phieu, "->", type(duoc_niem_phieu))
````

**Kiểm tra loại dữ liệu:**

Dùng hàm `type()` để xem loại của một biến:

```python
type(100)       # → int
type(75.5)      # → float
type("VNM")    # → str
type(True)      # → bool
```

## 3. Phép toán cơ bản

**Giải thích:**

Python có thể làm các phép toán như máy tính:

| Phép toán | Ký hiệu | Ví dụ | Kết quả |
|-----------|---------|-------|--------|
| Cộng | `+` | `10 + 5` | `15` |
| Trừ | `-` | `10 - 5` | `5` |
| Nhân | `*` | `10 * 5` | `50` |
| Chia | `/` | `10 / 5` | `2.0` |
| Chia lấy dư | `%` | `10 % 3` | `1` |
| Lũy thừa | `**` | `10 ** 2` | `100` |

````python
# =============================================================================
# 3. Phép toán cơ bản
# =============================================================================

# Ví dụ: Tính lợi nhuận cổ phiếu
gia_mua = 100000  # Giá mua: 100,000 VND
gia_ban = 120000  # Giá bán: 120,000 VND

# Tính lợi nhuận tuyệt đối
loi_nhuan = gia_ban - gia_mua
print(f"Lợi nhuận tuyệt đối: {loi_nhuan:,} VND")

# Tính tỷ suất lợi nhuận (%)
ty_suat = (gia_ban - gia_mua) / gia_mua
print(f"Tỷ suất lợi nhuận: {ty_suat:.2%}")  # .2% = hiển thị 2 chữ số thập phân

# Tính giá trị sau khi tăng 10%
gia_sau_tang = gia_mua * 1.10
print(f"Giá sau khi tăng 10%: {gia_sau_tang:,.0f} VND")
````

**Mẹo format số:**

- `{:,.0f}` → hiển thị số có dấu phẩy: `120,000`
- `{:.2%}` → hiển thị phần trăm: `0.20` → `20.00%`
- `{:.2f}` → hiển thị 2 số thập phân: `3.14`

## 4. List (Danh sách) - Lưu nhiều giá trị

**Giải thích:**

List giống như một **danh sách mua sắm**. Bạn có thể lưu nhiều thứ trong một biến.

```python
# Tạo một list (danh sách)
danh_sach = ["táo", "cam", "chuối"]

# List có thể chứa số
gia_cac_loai = [75000, 45000, 35000]
```

````python
# =============================================================================
# 4. List (Danh sách)
# =============================================================================

# Ví dụ 1: Danh sách các mã cổ phiếu
cac_ma_cp = ["VNM", "HPG", "VIC", "MSN", "VHM"]
print("Danh sách mã CP:", cac_ma_cp)

# Ví dụ 2: Danh sách giá
gia_cac_cp = [75000, 35000, 45000, 85000, 38000]
print("Giá các CP:", gia_cac_cp)

# --- Lấy phần tử từ List ---
# CHỈ SỐ (INDEX) bắt đầu từ 0!
print()
print("Phần tử đầu tiên:", cac_ma_cp[0])   # VNM
print("Phần tử thứ 2:", cac_ma_cp[1])    # HPG
print("Phần tử cuối cùng:", cac_ma_cp[-1]) # VHM
````

**QUAN TRỌNG: Chỉ số bắt đầu từ 0**

```
Danh sách:    ["A",    "B",    "C",    "D"]
Chỉ số:        0      1      2      3
              -4     -3     -2     -1   (đếm từ cuối)
```

Đây là quy ước của hầu hết ngôn ngữ lập trình. Dễ nhầm lẫn lúc đầu, quen sau.

````python
# --- Các thao tác với List ---

# Thêm phần tử vào cuối
cac_ma_cp.append("FPT")
print("Sau khi thêm FPT:", cac_ma_cp)

# Đếm số phần tử
print("Số lượng mã CP:", len(cac_ma_cp))

# Tìm vị trí của phần tử
vi_tri = cac_ma_cp.index("HPG")
print("HPG ở vị trí số:", vi_tri)

# Kiểm tra có trong list không
co_fpt = "FPT" in cac_ma_cp
co_xyz = "XYZ" in cac_ma_cp
print(f"Có FPT trong danh sách? {co_fpt}")  # True
print(f"Có XYZ trong danh sách? {co_xyz}")  # False
````

## 5. Vòng lặp (Loops) - Lặp đi lặp lại

**Giải thích:**

Vòng lặp giúp bạn **làm cùng một việc** nhiều lần.

Ví dụ trong đời thực: "Gửi tin nhắn chúc mừng đến 100 khách hàng"

Trong Python, có 2 loại vòng lặp:
- `for` — khi biết trước số lần lặp
- `while` — khi chỉ biết điều kiện dừng

````python
# =============================================================================
# 5. Vòng lặp (Loops)
# =============================================================================

# --- for loop: Lặp qua từng phần tử ---
print("=== Ví dụ 1: Duyệt danh sách ===")
cac_ma_cp = ["VNM", "HPG", "VIC", "MSN"]

for ma_cp in cac_ma_cp:
    print(f"- {ma_cp}")

# Giải thích từng bước:
# Bước 1: ma_cp = "VNM", in ra "- VNM"
# Bước 2: ma_cp = "HPG", in ra "- HPG"
# Bước 3: ma_cp = "VIC", in ra "- VIC"
# Bước 4: ma_cp = "MSN", in ra "- MSN"
# Hết danh sách → dừng
````

````python
# --- for loop với range() ---
print("\n=== Ví dụ 2: Lặp 5 lần ===")

for i in range(5):
    print(f"Lần thứ {i + 1}")

# range(5) tạo ra: 0, 1, 2, 3, 4
# Cộng thêm 1 để hiển thị: 1, 2, 3, 4, 5
````

````python
# --- while loop: Lặp đến khi điều kiện sai ---
print("\n=== Ví dụ 3: While loop ===")

so_du = 100000  # Số dư tài khoản
toc_do_tang = 1.10  # Tăng 10% mỗi năm
nam = 0

while so_du < 200000:  # Lặp đến khi đạt 200,000
    nam = nam + 1
    so_du = so_du * toc_do_tang
    print(f"Năm {nam}: {so_du:,.0f} VND")

print(f"\n→ Sau {nam} năm, số dư đạt {so_du:,.0f} VND")
````

## 6. Hàm (Function) - Tái sử dụng code

**Giải thích:**

Hàm giống như một **công thức nấu ăn**. Bạn viết một lần, dùng nhiều lần.

```python
def ten_ham(tham_so):

```

````python
# =============================================================================
# 6. Hàm (Function)
# =============================================================================

# --- Ví dụ 1: Hàm tính lợi nhuận ---
def tinh_loi_nhuan(gia_mua, gia_ban):
    """
    Tính lợi nhuận và tỷ suất lợi nhuận
    
    Args:
        gia_mua: Giá mua (VND)
        gia_ban: Giá bán (VND)
    
    Returns:
        Tuple (lợi nhuận tuyệt đối, tỷ suất %)
    """
    loi_nhuan = gia_ban - gia_mua
    ty_suat = (gia_ban - gia_mua) / gia_mua
    return loi_nhuan, ty_suat

# Sử dụng hàm
ln1, ts1 = tinh_loi_nhuan(100000, 120000)
print(f"Mua 100k, Bán 120k: Lợi nhuận = {ln1:,} VND ({ts1:.1%})")

ln2, ts2 = tinh_loi_nhuan(75000, 69000)
print(f"Mua 75k, Bán 69k: Lợi nhuận = {ln2:,} VND ({ts2:.1%})")
````

````python
# --- Ví dụ 2: Hàm tính giá trị danh mục ---
def tinh_gia_tri_danh_muc(hang_hoa):
    """
    Tính tổng giá trị danh mục đầu tư
    
    Args:
        hang_hoa: Dict chứa thông tin {tên: {giá, số lượng}}
    
    Returns:
        Tổng giá trị
    """
    tong = 0
    for ten, thong_tin in hang_hoa.items():
        gia = thong_tin["gia"]
        so_luong = thong_tin["so_luong"]
        gia_tri = gia * so_luong
        tong = tong + gia_tri
        print(f"{ten}: {gia:,} x {so_luong} = {gia_tri:,} VND")
    
    return tong

# Tạo danh mục đầu tư
danh_muc = {
    "VNM": {"gia": 75000, "so_luong": 100},
    "HPG": {"gia": 35000, "so_luong": 200},
    "VIC": {"gia": 45000, "so_luong": 50}
}

print("=== Tính giá trị danh mục ===")
tong_gia_tri = tinh_gia_tri_danh_muc(danh_muc)
print(f"\nTổng giá trị: {tong_gia_tri:,} VND")
````

## 7. Dictionary - Lưu dữ liệu dạng cặp Key-Value

**Giải thích:**

Dictionary giống như một **danh bạ điện thoại**:
- Key = Tên người (định danh)
- Value = Số điện thoại (thông tin)

```python
danh_ba = {
    "Minh": "0901234567",
    "Lan": "0912345678"
}
```

````python
# =============================================================================
# 7. Dictionary
# =============================================================================

# --- Ví dụ: Thông tin cổ phiếu VNM ---
thong_tin_vnm = {
    "ma": "VNM",
    "ten": "Vinamilk",
    "gia": 75000,
    "khoi_luong": 5_000_000,
    "san": "HOSE"
}

# Lấy thông tin
print("Mã:", thong_tin_vnm["ma"])
print("Tên:", thong_tin_vnm["ten"])
print("Giá:", thong_tin_vnm["gia"], "VND")

# Thêm thông tin mới
thong_tin_vnm["pe"] = 15.5  # P/E ratio
thong_tin_vnm["duoc_niem_phieu"] = True

print("P/E:", thong_tin_vnm["pe"])
print("Niêm yết:", thong_tin_vnm["duoc_niem_phieu"])
````

````python
# --- Ví dụ: Nhiều cổ phiếu ---
danh_sach_cp = {
    "VNM": {"gia": 75000, "kl": 5_000_000, "nganh": "Thực phẩm"},
    "HPG": {"gia": 35000, "kl": 10_000_000, "nganh": "Thép"},
    "VIC": {"gia": 45000, "kl": 3_000_000, "nganh": "Bất động sản"}
}

# Duyệt từng cổ phiếu
print("\n=== Danh sách cổ phiếu ===")
for ma, tt in danh_sach_cp.items():
    gia_tri = tt["gia"] * tt["kl"]
    print(f"{ma}: {tt['nganh']} - GT: {gia_tri:,.0f} VND")
````

## 8. Đọc và Ghi File

**Giải thích:**

File là nơi lưu dữ liệu. Chúng ta cần:
- **Đọc (Read)** — lấy dữ liệu từ file
- **Ghi (Write)** — lưu dữ liệu vào file

````python
# =============================================================================
# 8. Đọc và Ghi File
# =============================================================================

import json  # Thư viện xử lý JSON
from pathlib import Path  # Quản lý đường dẫn

# --- Ghi dữ liệu vào file ---

# Tạo dữ liệu mẫu
du_lieu = {
    "ngay": "2024-01-15",
    "danh_muc": [
        {"ma": "VNM", "gia": 75000, "so_luong": 100},
        {"ma": "HPG", "gia": 35000, "so_luong": 200}
    ]
}

# Đường dẫn file
duong_dan = Path("../data/danh_muc.json")
duong_dan.parent.mkdir(parents=True, exist_ok=True)

# Ghi file (như viết vào sổ)
with open(duong_dan, "w", encoding="utf-8") as f:
    json.dump(du_lieu, f, indent=2, ensure_ascii=False)

print(f"Đã lưu vào: {duong_dan}")
````

````python
# --- Đọc dữ liệu từ file ---

# Đọc file (như mở sổ ra đọc)
with open(duong_dan, "r", encoding="utf-8") as f:
    du_lieu_doc = json.load(f)

print(f"Ngày: {du_lieu_doc['ngay']}")
print(f"Số cổ phiếu: {len(du_lieu_doc['danh_muc'])}")

for cp in du_lieu_doc["danh_muc"]:
    gia_tri = cp["gia"] * cp["so_luong"]
    print(f"  - {cp['ma']}: {gia_tri:,} VND")
````

**Mẹo:**

```python
# Luôn dùng "with open()" để đọc/ghi file
# "with" đảm bảo file được đóng lại đúng cách
# encoding="utf-8" quan trọng để xử lý tiếng Việt
```
