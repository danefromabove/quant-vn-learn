## 1. BIẾN

**1. Khái niệm:** Nhãn gắn với giá trị
**2. Cách dùng:** `ten = gia_tri`
**3. Đơn giản:** Giống như dán nhãn lên hộp đựng đồ

```python
# Ví dụ: Lưu thông tin cổ phiếu
ma_cp = "VNM"
gia = 75000
print(f"{ma_cp}: {gia:,} VND")
```

## 2. KIỂU DỮ LIỆU

**1. Khái niệm:** Loại giá trị (số, chữ, đúng/sai)
**2. Cách dùng:** `int`, `float`, `str`, `bool`
**3. Đơn giản:** Như các loại đồ đựng khác nhau: hộp, túi, chai

```python
# Ví dụ: Các kiểu dữ liệu
so_nguyen = 100          # int: 1, 2, 3...
so_thap_phan = 75.5     # float: 1.5, 3.14...
chu = "VNM"              # str: "text"
la_co_phieu = True       # bool: True/False

print(f"{type(so_nguyen).__name__}: {so_nguyen}")
print(f"{type(so_thap_phan).__name__}: {so_thap_phan}")
print(f"{type(chu).__name__}: {chu}")
print(f"{type(la_co_phieu).__name__}: {la_co_phieu}")
```

## 3. PHÉP TOÁN

**1. Khái niệm:** Cộng, trừ, nhân, chia
**2. Cách dùng:** `+`, `-`, `*`, `/`, `%`, `**`
**3. Đơn giản:** Như máy tính bỏ túi

```python
# Ví dụ: Tính lợi nhuận
gia_mua = 100000
gia_ban = 120000

loi_nhuan = gia_ban - gia_mua          # Trừ
ty_suat = loi_nhuan / gia_mua         # Chia

print(f"Lợi nhuận: {loi_nhuan:,} VND")
print(f"Tỷ suất: {ty_suat:.2%}")
```

## 4. LIST

**1. Khái niệm:** Danh sách nhiều giá trị
**2. Cách dùng:** `ds = [a, b, c]`
**3. Đơn giản:** Giống danh sách mua sắm

```python
# Ví dụ: Danh sách cổ phiếu
cac_ma = ["VNM", "HPG", "VIC"]
cac_gia = [75000, 35000, 45000]

print(f"Mã đầu tiên: {cac_ma[0]}")  # Index bắt đầu từ 0
print(f"Giá HPG: {cac_gia[1]}")
```

## 5. VÒNG LẶP

**1. Khái niệm:** Lặp lại nhiều lần
**2. Cách dùng:** `for x in ds:` hoặc `while dk:`
**3. Đơn giản:** Như gửi tin nhắn cho 100 người

```python
# Ví dụ: Duyệt danh sách
cac_ma = ["VNM", "HPG", "VIC"]

for ma in cac_ma:
    print(f"- {ma}")
```

## 6. HÀM

**1. Khái niệm:** Khối code tái sử dụng
**2. Cách dùng:** `def ten(para): return ket_qua`
**3. Đơn giản:** Như công thức nấu ăn

```python
# Ví dụ: Hàm tính lợi nhuận
def tinh_loi_nhuan(mua, ban):
    return ban - mua

ln1 = tinh_loi_nhuan(100000, 120000)
ln2 = tinh_loi_nhuan(35000, 40000)
print(f"Lợi nhuận 1: {ln1:,} VND")
print(f"Lợi nhuận 2: {ln2:,} VND")
```

## 7. DICTIONARY

**1. Khái niệm:** Cặp key-value
**2. Cách dùng:** `d = {key: value}`
**3. Đơn giản:** Như danh bạ điện thoại

```python
# Ví dụ: Thông tin cổ phiếu
vnm = {
    "ma": "VNM",
    "ten": "Vinamilk",
    "gia": 75000
}

print(f"Mã: {vnm['ma']}")
print(f"Giá: {vnm['gia']:,} VND")
```

## 8. FILE

**1. Khái niệm:** Lưu/trữ dữ liệu
**2. Cách dùng:** `open()`, `json.dump()`
**3. Đơn giản:** Như viết sổ

```python
import json
from pathlib import Path

# Ghi file
du_lieu = {"ma": "VNM", "gia": 75000}
Path("../data").mkdir(exist_ok=True)
with open("../data/cp.json", "w", encoding="utf-8") as f:
    json.dump(du_lieu, f, indent=2)

# Đọc file
with open("../data/cp.json", "r", encoding="utf-8") as f:
    doc = json.load(f)
print(f"Đã đọc: {doc}")
```

# 🎯 BÀI TOÁN LỚN

**Yêu cầu:** Xây dựng chương trình quản lý danh mục cổ phiếu

Chương trình cần:
1. Lưu danh sách cổ phiếu (mã, giá mua, số lượng)
2. Tính tổng giá trị danh mục
3. Tính lợi nhuận của từng cổ phiếu
4. Lưu kết quả ra file JSON

```python
# =============================================================================
# GIẢI BÀI TOÁN: Quản lý danh mục cổ phiếu
# =============================================================================

# --- Bước 1: Tạo dữ liệu (BIẾN, LIST, DICTIONARY) ---

danh_muc = [
    {"ma": "VNM", "gia_mua": 72000, "sl": 100},
    {"ma": "HPG", "gia_mua": 32000, "sl": 200},
    {"ma": "VIC", "gia_mua": 48000, "sl": 50},
]

# Giá thị trường hiện tại (BIẾN)
gia_thi_truong = {
    "VNM": 75000,
    "HPG": 35000,
    "VIC": 45000,
}

print("=== DANH MỤC ĐẦU TƯ ===")
print(f"{'Mã':<6} {'Giá mua':>12} {'Giá hiện':>12} {'SL':>6} {'P/L':>12}")
print("-" * 55)
```

```python
# --- Bước 2: Duyệt và tính toán (VÒNG LẶP, PHÉP TOÁN) ---

tong_dau_tu = 0
tong_gia_tri = 0

for cp in danh_muc:
    ma = cp["ma"]
    gia_mua = cp["gia_mua"]
    sl = cp["sl"]
    gia_hien = gia_thi_truong[ma]
    
    # Tính P/L (PHÉP TOÁN)
    gia_tri_mua = gia_mua * sl
    gia_tri_hien = gia_hien * sl
    pl = gia_tri_hien - gia_tri_mua
    pl_pct = pl / gia_tri_mua
    
    # Cộng dồn (VÒNG LẶP)
    tong_dau_tu += gia_tri_mua
    tong_gia_tri += gia_tri_hien
    
    print(f"{ma:<6} {gia_tri_mua:>12,} {gia_tri_hien:>12,} {sl:>6} {pl:>+12,} ({pl_pct:+.1%})")

print("-" * 55)
```

```python
# --- Bước 3: Tổng kết (HÀM) ---

def in_tong_ket(tong_dau_tu, tong_gia_tri):
    pl_tong = tong_gia_tri - tong_dau_tu
    pl_pct = pl_tong / tong_dau_tu
    
    print(f"{'TỔNG':<6} {tong_dau_tu:>12,} {tong_gia_tri:>12,} {'':<6} {pl_tong:>+12,} ({pl_pct:+.1%})")
    print()
    print(f"💰 Tổng đầu tư: {tong_dau_tu:,} VND")
    print(f"💵 Giá trị hiện tại: {tong_gia_tri:,} VND")
    print(f"📈 Lợi nhuận: {pl_tong:+,} VND ({pl_pct:+.2%})")
    return {"tong_dau_tu": tong_dau_tu, "tong_gia_tri": tong_gia_tri, "loi_nhuan": pl_tong}

ket_qua = in_tong_ket(tong_dau_tu, tong_gia_tri)
```

```python
# --- Bước 4: Lưu kết quả (FILE) ---

import json
from pathlib import Path

# Chuẩn bị dữ liệu xuất
ket_qua_xuat = {
    "ngay": "2024-01-15",
    "danh_muc": danh_muc,
    "tong_dau_tu": tong_dau_tu,
    "tong_gia_tri": tong_gia_tri,
    "loi_nhuan": tong_gia_tri - tong_dau_tu
}

# Lưu file
duong_dan = Path("../data/danh_muc_2024.json")
duong_dan.parent.mkdir(parents=True, exist_ok=True)

with open(duong_dan, "w", encoding="utf-8") as f:
    json.dump(ket_qua_xuat, f, indent=2, ensure_ascii=False)

print(f"\n✅ Đã lưu vào: {duong_dan}")
```
