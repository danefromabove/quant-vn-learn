# 🎯 PROBLEM 1: Tính Lợi Nhuận Danh Mục

**Yêu cầu:** Viết chương trình tính lợi nhuận cho danh mục cổ phiếu

**Input:**
- Một danh sách cổ phiếu, mỗi cổ phiếu có: mã, giá mua, số lượng
- Giá thị trường hiện tại của từng mã

**Output:**
- Với mỗi cổ phiếu: mã, giá trị mua, giá trị hiện tại, lợi nhuận (VNĐ), lợi nhuận (%)
- Tổng giá trị mua của toàn danh mục
- Tổng giá trị hiện tại của toàn danh mục
- Tổng lợi nhuận của toàn danh mục

**Ràng buộc:**
- Sử dụng vòng lặp để duyệt danh sách
- Sử dụng hàm để tính lợi nhuận cho một cổ phiếu
- Sử dụng dictionary để lưu thông tin cổ phiếu

# 🎯 PROBLEM 2: Phân Tích Top Cổ Phiếu

**Yêu cầu:** Tìm và phân tích top N cổ phiếu tốt nhất

**Input:**
- Một danh sách 20 cổ phiếu, mỗi cổ phiếu có: mã, giá, P/E, khối lượng giao dịch
- Số N (ví dụ: top 5)

**Output:**
- Top N cổ phiếu có giá cao nhất
- Top N cổ phiếu có P/E thấp nhất
- Top N cổ phiếu có khối lượng giao dịch lớn nhất
- Trung bình giá, trung bình P/E của toàn bộ danh sách

**Ràng buộc:**
- Dùng vòng lặp để tìm top N
- Dùng hàm để sắp xếp và lọc
- Dùng dictionary để nhóm dữ liệu theo tiêu chí

# 🎯 PROBLEM 3: Tính Giá Trị Danh Mục Qua Thời Gian

**Yêu cầu:** Mô phỏng giá trị danh mục thay đổi theo thời gian

**Input:**
- Một danh mục gồm nhiều cổ phiếu (mã, số lượng)
- Giá ban đầu của mỗi cổ phiếu
- Tỷ lệ tăng trưởng hàng năm của mỗi cổ phiếu (mảng N năm)
- Số năm mô phỏng

**Output:**
- Bảng giá trị danh mục qua từng năm (Year 0, Year 1, ..., Year N)
- Năm nào danh mục đạt giá trị gấp đôi
- Cổ phiếu nào đóng góp nhiều nhất vào lợi nhuận

**Ràng buộc:**
- Dùng vòng lặp lồng nhau (năm → cổ phiếu)
- Dùng hàm để tính giá trị một cổ phiếu qua N năm
- Dùng dictionary để lưu giá trị theo năm và theo cổ phiếu
