## 1. Variables & Data Types / Biến & Kiểu dữ liệu

**Giải thích / Explanation:**

Trong Python, mọi thứ đều là object. Variables là labels trỏ đến objects.

*In Python, everything is an object. Variables are labels pointing to objects.*

```python
# =============================================================================
# 1. Variables & Data Types / Biến & Kiểu dữ liệu
# =============================================================================

# --- Numbers / Số ---
# Integer (số nguyên)
age = 25
print(f"Age: {age}, Type: {type(age)}")

# Float (số thực)
price = 150000.50
print(f"Price: {price}, Type: {type(price)}")

# --- Strings / Chuỗi ---
company = "VIC"
company_name = "Vingroup"
full_name = f"{company} - {company_name}"  # f-string formatting
print(f"Company: {full_name}")

# --- Booleans ---
is_listed = True
is_bankrupt = False
print(f"Is VIC listed? {is_listed}")
```

**💡 Tips / Mẹo:**

- Python sử dụng **snake_case** cho variable names: `stock_price`, `market_cap`
- **f-strings** (f"...") là cách hiện đại để format strings
- **Type hints** giúp code rõ ràng hơn (Python 3.5+)

```python
# Type hints / Gợi ý kiểu (Python 3.13+)
def calculate_return(initial: float, final: float) -> float:
    """Calculate simple return / Tính lợi nhuận đơn giản"""
    return (final - initial) / initial

stock_return = calculate_return(100000, 120000)
print(f"Return: {stock_return:.2%}")  # 20.00%
```

## 2. Lists & List Comprehensions

**Giải thích / Explanation:**

List là collection có thứ tự, có thể thay đổi (mutable).

*List is an ordered, mutable collection.*

```python
# =============================================================================
# 2. Lists / Danh sách
# =============================================================================

# Tạo list / Create a list
vn_indices = ["VN-Index", "HNX-Index", "VN30"]

# Indexing (0-based) / Chỉ mục
print(f"First index: {vn_indices[0]}")
print(f"Last index: {vn_indices[-1]}")

# Slicing / Cắt list
prices = [100, 105, 103, 108, 110, 107]
print(f"First 3 prices: {prices[:3]}")
print(f"Last 2 prices: {prices[-2:]}")

# List methods / Phương thức
prices.append(115)  # Thêm vào cuối
prices.remove(103)  # Xóa giá trị
print(f"Modified prices: {prices}")
```

```python
# --- List Comprehension / Biểu thức danh sách ---
# Cú pháp ngắn gọn để tạo list

# Ví dụ: Tính lợi nhuận cho nhiều mã cổ phiếu
initial_prices = [100, 200, 150, 80]
final_prices = [110, 180, 165, 85]

# Cách 1: Loop truyền thống
returns_loop = []
for i in range(len(initial_prices)):
    ret = (final_prices[i] - initial_prices[i]) / initial_prices[i]
    returns_loop.append(ret)
print(f"Returns (loop): {returns_loop}")

# Cách 2: List comprehension
returns_comp = [(final - initial) / initial for initial, final in zip(initial_prices, final_prices)]
print(f"Returns (comprehension): {returns_comp}")
```

**💡 Tips:**

- List comprehension **nhanh hơn** loop thông thường
- **zip()** ghép 2+ lists lại với nhau
- Dùng **enumerate()** khi cần cả index và value

## 3. Dictionaries / Từ điển

**Giải thích / Explanation:**

Dict (dictionary) lưu trữ key-value pairs. Rất hữu ích cho stock data.

*Dict stores key-value pairs. Very useful for stock data.*

```python
# =============================================================================
# 3. Dictionaries / Từ điển
# =============================================================================

# Tạo dict / Create dict
stock_data = {
    "symbol": "VNM",
    "name": "Vinamilk",
    "price": 75000,
    "volume": 5_000_000,
    "market_cap": 50_000_000_000_000  # 50 trillion VND
}

# Truy cập / Access
print(f"Symbol: {stock_data['symbol']}")
print(f"Price: {stock_data.get('price', 'N/A')}")

# Thêm/Sửa / Add/Edit
stock_data["sector"] = "Consumer Goods"
stock_data["price"] = 76000

# Dict comprehension
stock_prices = {"VNM": 75000, "VIC": 45000, "HPG": 35000}
premium_prices = {k: v * 1.1 for k, v in stock_prices.items()}  # 10% premium
print(f"Premium prices: {premium_prices}")
```

**💡 Tips:**

- Dùng **.get()** để tránh KeyError khi key không tồn tại
- **Dict comprehensions** tương tự list comprehensions
- **items()** trả về cặp key-value để iterate

## 4. Control Flow / Luồng điều khiển

**Giải thích / Explanation:**

Control flow quyết định code chạy như thế nào: if/else, for, while.

*Control flow determines how code executes: if/else, for, while.*

```python
# =============================================================================
# 4. Control Flow / Luồng điều khiển
# =============================================================================

# --- If/Else ---
def classify_return(return_pct: float) -> str:
    """Classify stock return / Phân loại lợi nhuận"""
    if return_pct > 0.20:
        return "Excellent (Xuất sắc)"
    elif return_pct > 0.10:
        return "Good (Tốt)"
    elif return_pct > 0:
        return "Positive (Dương)"
    elif return_pct == 0:
        return "Neutral (Trung lập)"
    else:
        return "Negative (Âm)"

for ret in [0.25, 0.15, 0.05, 0, -0.10]:
    print(f"{ret:+.0%} -> {classify_return(ret)}")
```

```python
# --- For Loop ---
vn_stocks = ["VNM", "VIC", "HPG", "MSN", "VHM"]

# enumerate: lấy cả index và value
for i, symbol in enumerate(vn_stocks):
    print(f"{i+1}. {symbol}")

# --- While Loop ---
# Tìm số lần giá tăng gấp đôi
price = 100
target = 200
years = 0
while price < target:
    price *= 1.10  # 10% annual return
    years += 1
print(f"Price doubles after {years} years")
```

**💡 Tips:**

- Dùng **for** khi biết số lần lặp
- Dùng **while** khi điều kiện dừng không cố định
- **break** thoát loop sớm, **continue** bỏ qua iteration hiện tại

## 5. Functions / Hàm

**Giải thích / Explanation:**

Functions là reusable blocks of code. Giúp organize và reuse logic.

*Functions are reusable code blocks. Help organize and reuse logic.*

```python
# =============================================================================
# 5. Functions / Hàm
# =============================================================================

# --- Basic Function ---
def calculate_log_return(price_t0: float, price_t1: float) -> float:
    """
    Calculate log return / Tính lợi nhuận log
    
    Args:
        price_t0: Giá tại thời điểm t0
        price_t1: Giá tại thời điểm t1
    
    Returns:
        Log return (ln(P1/P0))
    """
    return math.log(price_t1 / price_t0)

import math

log_ret = calculate_log_return(100, 120)
print(f"Log return: {log_ret:.4f}")
print(f"Simple return: {math.exp(log_ret) - 1:.4f}")
```

```python
# --- Lambda Functions / Hàm Lambda ---
# Hàm ẩn danh, viết trong 1 dòng

# Lambda cơ bản
square = lambda x: x ** 2
print(f"Square of 5: {square(5)}")

# Lambda với nhiều tham số
calculate_sharpe = lambda ret, vol, risk_free: (ret - risk_free) / vol
sharpe = calculate_sharpe(0.15, 0.20, 0.05)
print(f"Sharpe Ratio: {sharpe:.2f}")

# Lambda với sorting
stocks = [
    {"symbol": "VNM", "price": 75, "pe": 15},
    {"symbol": "HPG", "price": 35, "pe": 8},
    {"symbol": "VIC", "price": 45, "pe": 25}
]
sorted_by_pe = sorted(stocks, key=lambda x: x["pe"])
print("Sorted by P/E:", [s["symbol"] for s in sorted_by_pe])
```

**💡 Tips:**

- Dùng **docstrings** (triple quotes) để document functions
- **Type hints** giúp code dễ đọc và debug
- **Lambda** dùng cho simple operations, không thay thế được def cho phức tạp

## 6. File I/O & JSON

**Giải thích / Explanation:**

Đọc/ghi files là kỹ năng quan trọng cho data work.

*Reading/writing files is essential for data work.*

```python
# =============================================================================
# 6. File I/O & JSON
# =============================================================================

import json
from pathlib import Path

# --- Write JSON ---
stock_portfolio = {
    "investor": "Quant Learner",
    "holdings": [
        {"symbol": "VNM", "shares": 100, "avg_price": 72000},
        {"symbol": "HPG", "shares": 200, "avg_price": 32000}
    ],
    "cash": 5_000_000
}

# Save to file
output_path = Path("../data/portfolio.json")
output_path.parent.mkdir(parents=True, exist_ok=True)
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(stock_portfolio, f, indent=2, ensure_ascii=False)
print(f"Saved portfolio to {output_path}")

# --- Read JSON ---
with open(output_path, "r", encoding="utf-8") as f:
    loaded_portfolio = json.load(f)

print(f"\nLoaded portfolio:")
print(f"  Investor: {loaded_portfolio['investor']}")
print(f"  Holdings: {len(loaded_portfolio['holdings'])} stocks")
print(f"  Cash: {loaded_portfolio['cash']:,.0f} VND")
```

**💡 Tips:**

- Dùng **Path** từ pathlib thay vì os.path (cleaner)
- **with open()** tự động close file sau khi xong
- **encoding="utf-8"** quan trọng cho tiếng Việt
- **ensure_ascii=False** giữ unicode characters

## 7. Working with Stock Data / Làm việc với dữ liệu cổ phiếu

**Practice Exercise / Bài tập thực hành:**

Áp dụng những gì đã học để xử lý dữ liệu cổ phiếu đơn giản.

*Apply what you've learned to process simple stock data.*

```python
# =============================================================================
# 7. Mini Project: Stock Portfolio Calculator
# =============================================================================

def calculate_portfolio_value(holdings: list[dict]) -> dict:
    """
    Calculate portfolio metrics / Tính các chỉ số portfolio
    
    Args:
        holdings: List of {symbol, shares, avg_price}
    
    Returns:
        Dict với total value, cost, unrealized P/L
    """
    total_cost = 0
    
    # Giá hiện tại giả định (trong thực tế lấy từ API)
    current_prices = {
        "VNM": 75000,
        "HPG": 35000,
        "VIC": 45000,
        "MSN": 85000
    }
    
    position_details = []
    
    for h in holdings:
        symbol = h["symbol"]
        shares = h["shares"]
        avg_price = h["avg_price"]
        current_price = current_prices.get(symbol, avg_price)  # fallback
        
        cost = shares * avg_price
        value = shares * current_price
        pnl = value - cost
        pnl_pct = pnl / cost * 100
        
        total_cost += cost
        
        position_details.append({
            "symbol": symbol,
            "shares": shares,
            "avg_price": avg_price,
            "current_price": current_price,
            "cost": cost,
            "value": value,
            "pnl": pnl,
            "pnl_pct": pnl_pct
        })
    
    total_value = sum(p["value"] for p in position_details)
    total_pnl = total_value - total_cost
    
    return {
        "positions": position_details,
        "summary": {
            "total_cost": total_cost,
            "total_value": total_value,
            "total_pnl": total_pnl,
            "total_pnl_pct": total_pnl / total_cost * 100
        }
    }

# Test / Kiểm tra
my_holdings = [
    {"symbol": "VNM", "shares": 100, "avg_price": 72000},
    {"symbol": "HPG", "shares": 200, "avg_price": 32000},
    {"symbol": "VIC", "shares": 50, "avg_price": 48000}
]

result = calculate_portfolio_value(my_holdings)

print("📊 Portfolio Analysis / Phân tích Portfolio")
print("=" * 60)
print(f"{'Symbol':<8} {'Shares':>8} {'Avg Price':>12} {'Current':>12} {'P/L %':>10}")
print("-" * 60)

for pos in result["positions"]:
    print(f"{pos['symbol']:<8} {pos['shares']:>8} {pos['avg_price']:>12,.0f} {pos['current_price']:>12,.0f} {pos['pnl_pct']:>+9.1f}%")

print("-" * 60)
summary = result["summary"]
print(f"{'TOTAL':<8} {'':<8} {'':<12} {summary['total_cost']:>12,.0f} {summary['total_pnl_pct']:>+9.1f}%")
print()
print(f"💰 Total Invested: {summary['total_cost']:,.0f} VND")
print(f"💵 Current Value: {summary['total_value']:,.0f} VND")
print(f"📈 Unrealized P/L: {summary['total_pnl']:+,.0f} VND ({summary['total_pnl_pct']:+.1f}%)")
```
