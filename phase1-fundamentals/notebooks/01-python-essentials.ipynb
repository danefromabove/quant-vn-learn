# 🎯 PROBLEM 1: Tính Lợi Nhuận Danh Mục

**Trọng tâm:** Hàm + Vòng lặp + Dictionary

**Đề bài:**

Cho danh sách 10 cổ phiếu trong danh mục đầu tư:

```
| Mã | Giá mua | Số lượng | Giá hiện tại |
|-----|---------|----------|---------------|
| VNM | 72,000 | 100 | 75,000 |
| HPG | 32,000 | 200 | 35,000 |
| VIC | 48,000 | 50 | 45,000 |
| ... | ... | ... | ... |
```

**Yêu cầu:**

1. Tính giá trị mua = Giá mua × Số lượng
2. Tính giá trị hiện tại = Giá hiện tại × Số lượng
3. Tính lợi nhuận tuyệt đối = Giá trị hiện tại - Giá trị mua
4. Tính lợi nhuận % = (Lợi nhuận / Giá trị mua) × 100
5. Tính tổng danh mục: tổng giá trị mua, tổng giá trị hiện tại, tổng lợi nhuận

**Output mong đợi:**

```
=== DANH MỤC ĐẦU TƯ ===
Mã      Giá trị mua    Giá trị hiện    Lợi nhuận
VNM      7,200,000       7,500,000       +300,000 (+4.2%)
HPG      6,400,000       7,000,000       +600,000 (+9.4%)
...
TỔNG:   XX,XXX,XXX      XX,XXX,XXX      +X,XXX,XXX (+X.X%)

Lợi nhuận tổng: X,XXX,XXX VND
```

**Gợi ý:**
- Dùng **dictionary** để lưu thông tin mỗi cổ phiếu
- Dùng **vòng lặp** để duyệt qua từng cổ phiếu
- Dùng **hàm** để tính lợi nhuận của một cổ phiếu
- Dùng **phép toán** để tính các giá trị
- Dùng **biến** để lưu tổng

# 🎯 PROBLEM 2: Phân Tích Top Cổ Phiếu

**Trọng tâm:** Dictionary + Vòng lặp + Hàm

**Đề bài:**

Cho dữ liệu 20 cổ phiếu trên thị trường:

```
| Mã | Giá | P/E | KLGD (triệu) | Ngành |
|-----|------|-----|---------------|-------|
| VNM | 75,000 | 15.5 | 5.2 | Thực phẩm |
| HPG | 35,000 | 8.2 | 12.3 | Thép |
| VIC | 45,000 | 25.1 | 3.8 | BĐS |
| ... | ... | ... | ... | ... |
```

**Yêu cầu:**

1. Tìm **top 5 cổ phiếu có giá cao nhất**
2. Tìm **top 5 cổ phiếu có P/E thấp nhất** (cổ phiếu giá trị)
3. Tìm **top 5 cổ phiếu có khối lượng giao dịch lớn nhất**
4. Tính **trung bình giá** và **trung bình P/E** của toàn bộ
5. Đếm xem mỗi **ngành** có bao nhiêu cổ phiếu
6. Lưu kết quả phân tích ra **file JSON**

**Output mong đợi:**

```
=== TOP 5 THEO GIÁ ===
1. VHM - 120,000 VND
2. VIC - 95,000 VND
...

=== TOP 5 THEO P/E THẤP ===
1. HPG - 8.2
2. VPB - 9.1
...

=== THỐNG KÊ THEO NGÀNH ===
Thực phẩm: 3 cổ phiếu
Thép: 2 cổ phiếu
BĐS: 5 cổ phiếu
...
```

**Gợi ý:**
- Dùng **dictionary** để lưu thông tin từng cổ phiếu và thống kê ngành
- Dùng **vòng lặp** để duyệt và đếm
- Dùng **hàm** để tìm top N và tính trung bình
- Dùng **list** để lưu kết quả top N
- Dùng **if/else** để so sánh
- Dùng **file JSON** để lưu kết quả

# 🎯 PROBLEM 3: Mô Phỏng Danh Mục Qua N Năm

**Trọng tâm:** Vòng lặp + Hàm + Dictionary

**Đề bài:**

Bạn đầu tư vào danh mục cổ phiếu với tỷ lệ tăng trưởng khác nhau mỗi năm:

```
| Mã | Số lượng | Giá ban đầu | Tăng trưởng năm 1 | Tăng trưởng năm 2 | ... |
|-----|----------|--------------|-------------------|-------------------|-----|
| VNM | 100 | 72,000 | +5% | +8% | ... |
| HPG | 200 | 32,000 | +12% | -3% | ... |
| VIC | 50 | 48,000 | +15% | +10% | ... |
```

**Yêu cầu:**

1. Tính **giá trị danh mục** tại thời điểm ban đầu (Year 0)
2. Mô phỏng giá trị danh mục qua **5 năm**, mỗi năm áp dụng tỷ lệ tăng trưởng tương ứng
3. Tìm **năm nào danh mục đạt giá trị gấp đôi** so với ban đầu
4. Xác định **cổ phiếu nào đóng góp nhiều nhất** vào lợi nhuận cuối cùng
5. Vẽ bảng giá trị qua các năm:

```
=== GIÁ TRỊ DANH MỤC QUA CÁC NĂM ===
                Year 0    Year 1    Year 2    Year 3    Year 4    Year 5
VNM           7,200,000  7,560,000  8,165,000     ...       ...       ...
HPG           6,400,000  7,168,000  6,953,000     ...       ...       ...
VIC           2,400,000  2,760,000  3,036,000     ...       ...       ...
───────────────────────────────────────────────────────────────────────
TỔNG         16,000,000 17,488,000 18,154,000     ...       ...       ...

Năm danh mục gấp đôi: Year X
Cổ phiếu đóng góp nhiều nhất: VIC (+XXX,XXX VND)
```

**Gợi ý:**
- Dùng **dictionary lồng nhau** để lưu giá trị theo năm và theo cổ phiếu
- Dùng **vòng lặp lồng nhau**: vòng ngoài cho từng năm, vòng trong cho từng cổ phiếu
- Dùng **hàm** để tính giá trị một cổ phiếu sau N năm
- Dùng **phép toán** để cập nhật giá trị với tỷ lệ tăng trưởng
- Dùng **biến** để theo dõi năm gấp đôi và cổ phiếu tốt nhất
- Dùng **list** để lưu các năm
