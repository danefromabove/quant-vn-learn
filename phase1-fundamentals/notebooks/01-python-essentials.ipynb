### 3. DICTIONARY

**1. Định nghĩa:** Lưu dữ liệu dạng cặp key-value (khóa-giá trị)

**2. Cách dùng:**
```python
data = {"VNM": {"gia": 75000, "sl": 100}}
print(data["VNM"]["gia"])  # 75000
```

**3. Giải thích:** Như một ngăn tủ có nhiều ngăn nhỏ — mỗi ngăn có nhãn (key) và đồ bên trong (value). Muốn lấy gì, tra nhãn là thấy ngay.

**4. KHI NÀO DÙNG DICTIONARY vs LIST?**

| Dùng DICTIONARY khi... | Dùng LIST khi... |
|------------------------|-------------------|
| Cần tra cứu nhanh bằng tên (mã cổ phiếu) | Cần duyệt theo thứ tự |
| Mỗi phần tử có nhiều thuộc tính (giá, P/E, KL) | Chỉ cần lưu một danh sách đơn |
| Muốn truy cập ngẫu nhiên (random access) | Cần duyệt từ đầu đến cuối |
| Dữ liệu có cấu trúc phức tạp | Dữ liệu đơn giản, cùng loại |

**Ví dụ trong đầu tư:**
```python
# ✅ DICTIONARY - dữ liệu phức tạp, tra bằng tên
co_phieu = {
    "VNM": {"gia": 75000, "pe": 15.5, "klgd": 5.2},
    "HPG": {"gia": 35000, "pe": 8.2, "klgd": 12.3}
}
print(co_phieu["VNM"]["gia"])  # Tra nhanh bằng tên

# ✅ LIST - duyệt theo thứ tự, cần sorted
top_5 = sorted(co_phieu.items(), key=lambda x: x[1]["gia"], reverse=True)
# hoặc lưu danh sách ngày
ngay = ["2024-01", "2024-02", "2024-03"]
```

### 4. IF/ELSE (Điều Kiện)

**1. Định nghĩa:** Rẽ nhánh code dựa trên điều kiện đúng/sai

**2. Cách dùng:**
```python
if lai > 0:
    print("có lãi")
else:
    print("lỗ")

if pe < 15 and khoi_luong > 5:
    print("cổ phiếu tốt")
```

**3. Giải thích:** Như đèn giao thông — nếu xanh thì đi, đỏ thì dừng. Code cũng vậy: nếu điều kiện đúng thì làm A, sai thì làm B.

**4. IF/ELSE vs VÒNG LẶP: KHI NÀO DÙNG GÌ?**

| IF/ELSE (Rẽ nhánh) | VÒNG LẶP (Lặp) |
|---------------------|-------------------|
| Chọn MỘT đường đi trong nhiều đường | Làm cùng một việc NHIỀU lần |
| Hỏi: "Cái này thỏa mãn điều kiện không?" | Hỏi: "Còn phần tử nào chưa xử lý?" |
| Một lần quyết định | Nhiều lần lặp lại |
| Kết quả: thực hiện A HOẶC B | Kết quả: thực hiện cho TẤT CẢ |

**Ví dụ trong đầu tư:**
```python
# ✅ IF/ELSE - phân loại một cổ phiếu
if pe < 10:
    print("Rẻ")
elif pe < 20:
    print("Bình thường")
else:
    print("Đắt")

# ✅ VÒNG LẶP - xử lý tất cả cổ phiếu
for ma, cp in co_phieu.items():
    if cp["pe"] < 10:  # Kết hợp cả hai!
        print(f"{ma} rẻ")
```

**Quy tắc:**
- Dùng **IF/ELSE** khi cần **quyết định** cho một giá trị
- Dùng **VÒNG LẶP** khi cần **xử lý nhiều** giá trị
- **Kết hợp cả hai**: vòng lặp duyệt, if/else phân loại từng cái

# 🎯 PROBLEM 1: Tính Lợi Nhuận Danh Mục

**Trọng tâm:** Hàm + Vòng lặp + Dictionary

**Đề bài:**

Cho danh sách 10 cổ phiếu trong danh mục đầu tư:

```
| Mã | Giá mua | Số lượng | Giá hiện tại |
|-----|---------|----------|---------------|
| VNM | 72,000 | 100 | 75,000 |
| HPG | 32,000 | 200 | 35,000 |
| VIC | 48,000 | 50 | 45,000 |
| ... | ... | ... | ... |
```

**Yêu cầu:**

1. Tính giá trị mua = Giá mua × Số lượng
2. Tính giá trị hiện tại = Giá hiện tại × Số lượng
3. Tính lợi nhuận tuyệt đối = Giá trị hiện tại - Giá trị mua
4. Tính lợi nhuận % = (Lợi nhuận / Giá trị mua) × 100
5. Tính tổng danh mục: tổng giá trị mua, tổng giá trị hiện tại, tổng lợi nhuận

**Output mong đợi:**

```
=== DANH MỤC ĐẦU TƯ ===
Mã      Giá trị mua    Giá trị hiện    Lợi nhuận
VNM      7,200,000       7,500,000       +300,000 (+4.2%)
HPG      6,400,000       7,000,000       +600,000 (+9.4%)
...
TỔNG:   XX,XXX,XXX      XX,XXX,XXX      +X,XXX,XXX (+X.X%)

Lợi nhuận tổng: X,XXX,XXX VND
```

**Gợi ý:**
- Dùng **dictionary** để lưu thông tin mỗi cổ phiếu
- Dùng **vòng lặp** để duyệt qua từng cổ phiếu
- Dùng **hàm** để tính lợi nhuận của một cổ phiếu
- Dùng **phép toán** để tính các giá trị
- Dùng **biến** để lưu tổng

**Gợi ý:**
- Dùng **dictionary lồng nhau** để lưu giá trị theo năm và theo cổ phiếu
- Dùng **vòng lặp lồng nhau**: vòng ngoài cho từng năm, vòng trong cho từng cổ phiếu
- Dùng **hàm** để tính giá trị một cổ phiếu sau N năm
- Dùng **phép toán** để cập nhật giá trị với tỷ lệ tăng trưởng
- Dùng **biến** để theo dõi năm gấp đôi và cổ phiếu tốt nhất
- Dùng **list** để lưu các năm
